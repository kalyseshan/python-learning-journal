```python
class Node:
  def __init__(self, data):
    self.data = data
    self.left = None
    self.right = None
    self.height = 0 # existing node without children, or leaf, has a height = 0

class AVLTree:
  def __init__(self):
    self.root = None

  def getHeight(self, node):
    if not node:
      return -1
    return node.height

  def calculateHeight(self, node):
    if not node:
      return -1
    return max(self.getHeight(node.left), self.getHeight(node.right)) + 1

  def getBalance(self, node):
    if not node:
      return None
    return self.getHeight(node.left) - self.getHeight(node.right)

  def rightRotate(self, current):
    newRoot = current.left
    t2 = newRoot.right
    current.left = t2
    newRoot.right = current

    if current == self.root:
      self.root = newRoot

    current.height = self.calculateHeight(current)
    newRoot.height = self.calculateHeight(newRoot)
    return newRoot

  def leftRotate(self, current):
    newRoot = current.right
    t2 = newRoot.left
    current.right = t2
    newRoot.left = current

    if current == self.root:
      self.root = newRoot # reassigns root of tree after rotation

    current.height = self.calculateHeight(current)
    newRoot.height = self.calculateHeight(newRoot)
    return newRoot

  def traverse(self, node):
    if node:
      print(node.data)
      self.traverse(node.left)
      self.traverse(node.right)

  def leftRightRotate(self, current):
    current.left = self.leftRotate(current.left)
    return self.rightRotate(current)

  def rightLeftRotate(self, current):
    current.right = self.rightRotate(current.right)
    return self.leftRotate(current)

  def iterativeSearch(self, target):
    currentNode = self.root

    while currentNode != None:
      if currentNode.data == target:
        return currentNode
      elif currentNode.data < target:
        currentNode = currentNode.left
      else:
        currentNode = currentNode.right
    return None

  def recursiveSearch(self, current, target):
    # Does the current exist? if not then return False/None - indicates empty tree
    if current is None:
      return None
    # if it exists, is it == target?
    if target == current.data:
      return current
    # if target > current, search right
    elif target < current.data:
      return self.recursiveSearch(current.left, target)
    else:
      return self.recursiveSearch(current.right, target)

  def insert(self, current, data):
    # if current is none, then make a new node at this location
    if current is None:
      if not self.root:
        self.root = Node(data)
        return self.root
      else:
        return Node(data)

    # if not none, navigate to appropriate subtree
    elif data == current.data: # check for duplicate
      print("No duplicates allowed")
      return current
    elif data < current.data: # set left to recursive insert call of itself
      current.left = self.insert(current.left, data)
    else: # if the values are >=
      current.right = self.insert(current.right, data)

    print(current.data, "Height:", current.height)
    print("Right child: ", current.right)
    print("Left child: ", current.left)
    # Update height of node:
    current.height = self.calculateHeight(current)
    print(current.data, "Height:", current.height)

    # Balance new node
    balance = self.getBalance(current)
    print(current.data, "Balance:", balance)

    # Rotations
    if balance > 1:
      if current.left is not None and self.getBalance(current.left) < 0:
        return self.leftRightRotate(current)
      else:
        return self.rightRotate(current)

    if balance < -1:
      if current.right is not None and self.getBalance(current.right) < 0: # know it's
        return self.rightLeftRotate(current)
      else:
        return self.leftRotate(current)

    # return current node to ensure subtrees remain linked
    return current

  def delete(self, current, data):
    # Step 1: search for node we are deleting
    if current is None:
      return None
    elif data < current.data:
      current.left = self.delete(current.left, data)
    elif data > current.data:
      current.right = self.delete(current.right, data)
    else:
    # Step 2: Node found, check the case
    # Case 1: if node has no children, delete
      if current.left is None and current.right is None:
        return None
    # Case 2: if node has one child, point parent to that child
      elif current.left is None:
        return current.right
      elif current.right is None:
        return current.left

      else:
    # Case 3: Node has two children
      # find inorder successor (smallest in the right subtree)
        successor = self.inorderSuccessor(current.right)
      # replace data with inorder successor's data
        current.data = successor.data
      # recursively delete old inorder successor in its subtree
        current.right = self.delete(current.right, successor.data)

    # Update height of node:
    current.height = self.calculateHeight(current)

    # Balance new node
    balance = self.getBalance(current)

    # Rotations
    if balance > 1:
      if current.left is not None and self.getBalance(current.left) < 0:
        return self.leftRightRotate(current)
      elif current.right is not None and self.getBalance(current.right) < 0:
        return self.leftRightRotate(current)
      else:
        return self.rightRotate(current)

    if balance < -1:
      if current.left is not None and self.getBalance(current.left) < 0:
        return self.rightLeftRotate(current)
      elif current.right is not None and self.getBalance(current.right) < 0:
        return self.rightLeftRotate(current)
      else:
        return self.leftRotate(current)

  def inorderSuccessor(self, node):
    current = node
    while current.left is not None:
      current = current.leftRightRotate
    return current

  def inorderTraversal(self, node):
    if node:
      self.inorderTraversal(node.left)
      print(node.data)
      #print(node.getBalance)
      self.inorderTraversal(node.right)
```
